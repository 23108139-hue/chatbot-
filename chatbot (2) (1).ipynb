```python
from transformers import pipeline
import gradio as gr

generator = pipeline("text-generation", model="distilgpt2")

def chatbot(user_input):
    result = generator(
        user_input,
        max_length=100,
        do_sample=True,
        temperature=0.8,
        top_k=50,
        repetition_penalty=1.2
    )
    return result[0]["generated_text"]

demo = gr.Interface(
    fn=chatbot,
    inputs="text",
    outputs="text",
    title="My AI Chatbot"
)

demo.launch()
```

```python
from transformers import pipeline
import gradio as gr

generator = pipeline("text-generation", model="distilgpt2")

def chat(message, history):
    result = generator(
        message,
        max_length=120,
        do_sample=True,
        temperature=0.8,
        top_k=50,
        repetition_penalty=1.2
    )

    reply = result[0]["generated_text"].replace(message, "")
    return reply

demo = gr.ChatInterface(
    fn=chat,
    title="🤖 My AI Chatbot",
    description="Ask me anything!"
)

demo.launch()
```
